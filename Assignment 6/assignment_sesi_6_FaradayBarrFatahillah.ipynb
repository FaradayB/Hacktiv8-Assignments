**Task 1 — Fungsi kalkulasi dasar**
<br>Buat tiga fungsi berikut dalam satu cell:
1. hitung_subtotal(unit, harga_satuan): mengembalikan hasil perkalian unit dan harga_satuan.
2. hitung_diskon(subtotal, persen_diskon=0): mengembalikan nilai diskon dalam rupiah. Parameter persen_diskon adalah opsional dengan nilai default 0.
3. hitung_total(subtotal, persen_diskon=0): mengembalikan subtotal dikurangi diskon. Fungsi ini WAJIB memanggil fungsi hitung_diskon di dalamnya, bukan menghitung ulang sendiri.

<br>Uji ketiga fungsi dengan memanggil masing-masing minimal sekali dan cetak hasilnya.

````python
def format_angka(amount):
    return 'Rp {:,}'.format(amount)


def hitung_subtotal(unit, harga_satuan):
    return unit*harga_satuan


def hitung_diskon(subtotal, persen_diskon=0):
    return int(subtotal * persen_diskon/100)

def hitung_total(subtotal, persen_diskon=0):
    return subtotal - hitung_diskon(subtotal,persen_diskon)
````

**Task 2 — Fungsi laporan ringkas**
<br>Buat fungsi buat_laporan(data_penjualan, persen_diskon=0) yang:
1. Menerima list of dict penjualan dan persentase diskon (opsional).
2. Menghitung subtotal, diskon, dan total untuk setiap item menggunakan fungsi dari Task 1.
3. Mengembalikan sebuah dict berisi: "detail" (list of dict tiap item ditambah key subtotal, diskon, total), "grand_total" (total seluruh transaksi setelah diskon), dan "rata_unit" (rata-rata unit terjual per produk, float 2 desimal).

````python
def buat_laporan_data(data_penjualan, persentase_diskon=0):
    harga_total = 0
    grand_total = 0
    total_stok = 0
    list_detail = []
    detail_laporan = {}
    diskon = persentase_diskon
    for produk in data_penjualan:
        subtotal = hitung_subtotal(produk["unit"], produk["harga_satuan"])
        diskon = hitung_diskon(subtotal,persentase_diskon)
        harga_total = hitung_total(subtotal,persentase_diskon)

        total_stok += produk["unit"]
        grand_total += harga_total

        list_detail.append({**produk, "persentase_diskon":persentase_diskon, "subtotal":subtotal, "diskon":diskon, "total":harga_total})

    rata2_unit = round(total_stok/len(list_detail), 2)

    detail_laporan.update({
        "detail": list_detail,
        "grand_total": grand_total,
        "rata_unit" : rata2_unit
    })
    
    return detail_laporan

laporan = buat_laporan_data(penjualan)
laporan_diskon = buat_laporan_data(penjualan, 12)

print(f'Tanpa diskon:\nGrand Total: {laporan["grand_total"]}\nRata-rata penjualan unit: {laporan["rata_unit"]}\n')
print(f'Dengan diskon {laporan_diskon["detail"][0]["persentase_diskon"]}%:\nGrand Total: {laporan_diskon["grand_total"]}\nRata-rata penjualan unit: {laporan_diskon["rata_unit"]}')

# print(buat_laporan_data(penjualan))
# print(buat_laporan_data(penjualan, 12))    
````

**Task 3 — Fungsi dengan modul bawaan**

Buat fungsi tambahkan_timestamp(laporan) yang menerima dict laporan dari Task 2 dan menambahkan key baru "dibuat_pada" berisi string tanggal dan waktu saat fungsi dipanggil, dalam format: "27 April 2026, 14:30:05". Fungsi ini mengembalikan dict yang sama dengan key baru tersebut.


b. Menggunakan modul math:
Buat fungsi statistik_penjualan(data_penjualan) yang mengembalikan dict berisi: "total_unit", "max_subtotal", "min_subtotal", dan "std_subtotal". Hitung standar deviasi menggunakan rumus manual (bukan statistics.stdev), boleh memanfaatkan math.sqrt.

````python
#Task A
from datetime import datetime

def tambahkan_timestamp(laporan):
    bulan_indo = {"January": "Januari", 
                  "February": "Februari", 
                  "March": "Maret", 
                  "April": "April", 
                  "May": "Mei", 
                  "June": "Juni", 
                  "July": "Juli", 
                  "August": "Agustus", 
                  "September": "September", 
                  "October": "Oktober", 
                  "November": "November", 
                  "December": "Desember"}
   
    now = datetime.now()
    laporan["dibuat_pada"] = (f'{now.day} {bulan_indo[now.strftime('%B')]} {now.year}, {now.strftime('%H:%M:%S')}')

    return laporan

laporan_waktu = tambahkan_timestamp(laporan)
print(laporan_waktu['dibuat_pada'])

print(laporan["detail"][0]["unit"])
print(laporan["detail"][0]["harga_satuan"])



#Task B
import math 

def statistik_penjualan(data_penjualan):
    subtotals = []
    i = 0

    for data in data_penjualan:
        subtotal = hitung_subtotal(data["unit"], data["harga_satuan"])
        subtotals.append(subtotal)
        i+=1
    
    mean = sum(subtotals)/len(subtotals)
    variance = sum((x-mean) ** 2 for x in subtotals)/len(subtotals)
    std_subtotal = math.sqrt(variance)

    return {
        "max_subtotal": max(subtotals),
        "min_subtotal": min(subtotals),
        "std_subtotal": std_subtotal
    }

print(statistik_penjualan(penjualan))
````
