**Task 3 — Penanganan missing values**
1. Konversi kolom TotalCharges ke float. Baris yang tidak bisa dikonversi otomatis menjadi NaN. Simpan hasilnya ke kolom yang sama.
2. Isi missing values pada kolom TotalCharges menggunakan median. Jelaskan di markdown kenapa median lebih tepat dari mean untuk kolom ini.
3. Untuk kolom tenure dengan nilai 0: ganti dengan NaN, lalu isi menggunakan nilai median tenure pelanggan yang memiliki produk dan layanan yang sama (gunakan groupby berdasarkan InternetService dan PhoneService sebagai proxy).
4. Verifikasi: setelah cleaning, tidak ada lagi missing values di kolom TotalCharges dan tenure. Cetak konfirmasi dengan isnull().sum() untuk kedua kolom tersebut.

````python
#Task 3.1 
df['TotalCharges'] = df['TotalCharges'].astype('float')

df.info()
````

````python
df['TotalCharges'].value_counts(ascending=True)
````

````python
#Mean sebelum null value berubah menjadi median/mean
df['TotalCharges'].mean()
````

````python
#Task 3.2
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())

# Uncomment line di bawah dan comment line di atas kemudian run all untuk cek perubahan
# df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].mean())

df.query('TotalCharges_numeric.isna()')
````

Median lebih baik untuk data yang kurang simetris sehingga tidak bias ke rata-rata atau data dengan value yang tinggi

````python
#Mean setelah null value berubah menjadi median/mean
df['TotalCharges'].mean()
````

````python
df.query('tenure == 0.0')
````

````python
#Task 3.3
df.loc[df['tenure'] == 0, 'tenure'] = pd.NA

df['tenure'] = df.groupby(
    ['InternetService', 'PhoneService']
)['tenure'].transform(lambda x: x.fillna(x.median()))

df.query('TotalCharges_numeric.isna()')
````

````python
df = df.drop(columns=['TotalCharges_numeric'])
df.isna().sum().sort_values()
````
