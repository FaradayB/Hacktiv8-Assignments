**Task 1 — Class Karyawan (base class)**

Buat class Karyawan dengan ketentuan berikut:

Atribut (semua di-set saat instansiasi):
1. ````nama``` (public)
2. ```departemen``` (public)
3. ```__gaji_pokok``` (private, menggunakan name mangling)
 

Method:
1. ```__init__(self, nama, departemen, gaji_pokok)```: konstruktor.
2. ```get_gaji_pokok(self)```: mengembalikan nilai ```__gaji_pokok``` (getter).
3. ```set_gaji_pokok(self, nilai)```: memperbarui ```__gaji_pokok``` hanya jika nilai lebih besar dari 0. Kalau tidak, cetak pesan error.
4. ```hitung_gaji(self)```: mengembalikan __gaji_pokok apa adanya (akan di-override oleh subclass).
5. ```info(self)```: mencetak ringkasan karyawan dalam format: Nama, Departemen, Gaji Pokok, Total Gaji.
 

Uji class ini dengan membuat satu object dan memanggil method ```info()```.

`````python
def format_angka(amount):
    return 'Rp {:,}'.format(amount).replace(",",".")

class Karyawan:
    def __init__(self, nama:str, departemen:str, gaji_pokok):
        self.nama = nama
        self.departemen = departemen
        self.__gaji_pokok = gaji_pokok
    
    def get_gaji_pokok(self):
        return self.__gaji_pokok
    
    def set_gaji_pokok(self, nilai):
        if nilai > 0:
            self.__gaji_pokok = nilai
        else:
            print('Error: Tidak memasukkan gaji')
    
    def hitung_gaji(self):
        return self.__gaji_pokok
    
    def info(self):
        print(f'Nama          : {self.nama}')
        print(f'Departemen    : {self.departemen}')
        print(f'Gaji Pokok    : {format_angka(self.__gaji_pokok)}')
        print(f'Total Gaji    : {format_angka(self.hitung_gaji())}')


karyawan = Karyawan("Faraday Barr", "Trainee", 5000000000)
karyawan.info()
`````

**Task 2 — Subclass KaryawanTetap**
<br>Buat class KaryawanTetap yang mewarisi class Karyawan dengan ketentuan:

Atribut tambahan:
1. ```tunjangan_tetap``` (public): nilai rupiah tunjangan bulanan tetap.
2. ```bonus_tahunan``` (public): nilai rupiah bonus tahunan.
 

Method:
1. ```__init __```: panggil konstruktor parent menggunakan ```super()```, lalu set atribut tambahan.
2. ```hitung_gaji(self)```: mengembalikan ```gaji_pokok + tunjangan_tetap + (bonus_tahunan / 12)```. Pastikan hasilnya adalah integer.
3. Override method ```info()``` sehingga menampilkan tambahan: Jenis (Karyawan Tetap), Tunjangan, dan Bonus per bulan.
 

Buat dua object ```KaryawanTetap``` dan panggil ```info()``` pada keduanya.

`````python
class KaryawanTetap(Karyawan):
    def __init__(self, nama, departemen, gaji_pokok, tunjangan_tetap, bonus_tambahan):
        super().__init__(nama, departemen, gaji_pokok)
        self.tunjangan_tetap = tunjangan_tetap
        self.bonus_tambahan = bonus_tambahan
        self.jenis_karyawan = "Karyawan Tetap"

    def hitung_gaji(self):
        return int(self.get_gaji_pokok() + self.tunjangan_tetap + (self.bonus_tambahan/12))
    
    def info(self):
        print(f'Jenis Karyawan  : {self.jenis_karyawan}')
        print(f'Nama            : {self.nama}')
        print(f'Departemen      : {self.departemen}')
        print(f'Gaji Pokok      : {format_angka(self.get_gaji_pokok())}')
        print(f'Tunjangan       : {format_angka(self.tunjangan_tetap)}')
        print(f'Bonus per Bulan : {format_angka(int(self.bonus_tambahan/12))}')
        print(f'Total Gaji      : {format_angka(self.hitung_gaji())}')

kartap1 = KaryawanTetap("Faraday", "Junior SE", 10_000_000, 4_000_000, 60_000_000)
kartap2 = KaryawanTetap("Bobe", "Trainee", 6_000_000, 0, 0)

kartap1.info()
print('')
kartap2.info()
`````

**Task 3 — Subclass KaryawanKontrak**
<br>Buat class KaryawanKontrak yang mewarisi class Karyawan dengan ketentuan:

Atribut tambahan:
1. ```durasi_kontrak_bulan``` (public): lama kontrak dalam bulan.
2. ```rate_per_jam``` (public): tarif per jam dalam rupiah.
3. ```jam_per_bulan``` (public): jumlah jam kerja per bulan.
 

Method:
1. ```__init__```: panggil ```super()```, lalu set atribut tambahan.
2. ```hitung_gaji(self)```: mengembalikan ```rate_per_jam``` dikali ```jam_per_bulan```. Gaji pokok diabaikan untuk karyawan kontrak.
3. Override method ```info()``` sehingga menampilkan tambahan: Jenis (Karyawan Kontrak), Durasi Kontrak, Rate per jam, dan Jam per bulan.
 
Buat dua object ```KaryawanKontrak``` dan panggil ```info()``` pada keduanya.

`````python
class KaryawanKontrak(Karyawan):
    def __init__(self, nama, departemen, durasi_kontrak_bulan, rate_per_jam, jam_per_bulan, gaji_pokok=None):
        super().__init__(nama, departemen, gaji_pokok)
        self.durasi_kontrak_bulan = durasi_kontrak_bulan
        self.rate_per_jam = rate_per_jam
        self.jam_per_bulan = jam_per_bulan
        self.jenis_karyawan = "Karyawan Kontrak"
    
    def hitung_gaji(self):
        return int(self.rate_per_jam * self.jam_per_bulan)
    
    def info(self):
        print(f'Jenis Karyawan  : {self.jenis_karyawan}')
        print(f'Nama            : {self.nama}')
        print(f'Departemen      : {self.departemen}')
        print(f'Durasi Kontrak  : {self.durasi_kontrak_bulan}')
        print(f'Rate per Jam    : {self.rate_per_jam}')
        print(f'Jam per Bulan   : {self.jam_per_bulan}')
        print(f'Total Gaji      : {format_angka(self.hitung_gaji())}')

karkon = KaryawanKontrak("Faraday Barr", "SE", 25, 50000, 40)
karkon1 = KaryawanKontrak("Fatahillah", "SE", 32, 350_000, 40)
karkon.info()
print('')
karkon1.info()
`````

**Task 4 — Polymorphism dan manajemen list**
1. Buat list bernama ```tim_techspark``` yang berisi semua object yang sudah Anda buat di Task 2 dan Task 3 (total 4 karyawan).
2. Iterasi list tersebut menggunakan for loop. Untuk setiap karyawan, cetak nama, jenis class (gunakan ```type(obj).__name__```), dan hasil ```hitung_gaji()```. Perhatikan bahwa ```hitung_gaji()``` menghasilkan nilai berbeda tergantung jenis karyawan meski dipanggil dengan cara yang sama — inilah polymorphism.
3. Hitung dan cetak total pengeluaran gaji per bulan untuk seluruh karyawan dalam list tersebut.
4. Gunakan ```set_gaji_pokok()``` untuk menaikkan gaji pokok salah satu karyawan tetap sebesar 500.000. Cetak ```info()``` karyawan tersebut sebelum dan sesudah perubahan untuk membuktikan encapsulation berjalan.

`````python
kartap1 = KaryawanTetap("Faraday", "Junior SE", 10_000_000, 4_000_000, 60_000_000)
kartap2 = KaryawanTetap("Bobe", "Trainee", 6_000_000, 0, 0)
kartap3 = KaryawanTetap("Barr", "Senior SE", 14_000_000, 5_000_000, 72_000_000)
karkon1 = KaryawanKontrak("Fatahillah", "SE", 32, 350_000, 40)

tim_techspark = [kartap1, kartap2, kartap3, karkon1]

total_gaji = 0

print("Daftar Gaji Tim Techspark")
for kar in tim_techspark:
    gaji = kar.hitung_gaji()
    total_gaji += gaji
    print(f'Nama: {kar.nama} | Jenis: {type(kar).__name__} | Gaji: {format_angka(gaji)}')

print(f'\nTotal gaji: {format_angka(total_gaji)}')


print(f'\n\n======================\nSebelum kenaikan gaji')
kartap2.info()

kartap2.set_gaji_pokok(kartap2.get_gaji_pokok() + 500_000)

print(f'\nSetelah kenaikan gaji')
kartap2.info()
`````
