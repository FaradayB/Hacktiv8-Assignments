Anda diminta membangun skrip untuk memproses data transaksi harian sebuah kafe. Skrip ini menghitung total tagihan setiap pelanggan berdasarkan aturan diskon yang berlaku.

Gunakan data berikut:

```python
menu = {
    "Espresso": 25000,
    "Latte": 35000,
    "Cappuccino": 38000,
    "Croissant": 28000,
    "Sandwich": 42000,
    "Cheesecake": 45000,
}

transaksi = [
    {"nama": "Budi", "pesanan": ["Latte", "Croissant", "Croissant"]},
    {"nama": "Sari", "pesanan": ["Espresso", "Cheesecake"]},
    {"nama": "Dian", "pesanan": ["Cappuccino", "Sandwich", "Latte","Cheesecake"]},
    {"nama": "Reza", "pesanan": ["Espresso"]},
    {"nama": "Ayu",  "pesanan": ["Latte", "Latte", "Sandwich","Croissant", "Cheesecake"]},

]
```

**Task 1 — Hitung tagihan per pelanggan**
1. Iterasi list transaksi menggunakan for loop.
2. Untuk setiap pelanggan, hitung subtotal dari semua item pesanan dengan mengacu ke dict menu, tentukan persentase diskon, hitung jumlah diskon dalam rupiah, dan hitung total akhir.
3. Cetak ringkasan transaksi setiap pelanggan dalam format: Pelanggan: Budi Pesanan  : Latte, Croissant, 4. Croissant Subtotal : Rp 98.000 Diskon   : 10% (Rp 9.800) Total    : Rp 88.200 ----------------------------------------
4. Di akhir, cetak total pendapatan kafe dari semua transaksi setelah diskon.

```python
def hitung_transaksi(list_pesanan, menu):
    total = [menu[food] for food in list_pesanan]
    return sum(total)

def hitung_diskon(amount):
    if amount >= 100_000:
        diskon = 15
    elif amount >= 50_000:
        diskon = 10
    else:
        diskon = 0
    after_diskon = int(amount * (1 - (diskon/100)))
    return diskon, after_diskon

def format_angka(amount):
    return 'Rp {:,}'.format(amount)

def format_diskon(diskon):
    return '{:.2f}%'.format(diskon)

total_pendapatan_toko = 0
tagihan = []

print("Total per orang:")
for data in transaksi:
    nama = data["nama"]
    pesanan = data["pesanan"]

    total_perorang = hitung_transaksi(pesanan, menu)
    diskon, after_diskon = hitung_diskon(total_perorang)
    formatted_harga_prediskon = format_angka(total_perorang)
    formatted_harga_postdiskon = format_angka(after_diskon)
    tagihan.append(after_diskon)

    print(f'Pelanggan: {nama}\nPesanan: {", ".join(pesanan)}\nSubtotal: {formatted_harga_prediskon.replace(",",".")}\nDiskon: {format_diskon(diskon)} ({format_angka(total_perorang-after_diskon).replace(",",".")})\n------------------------------\nTotal: {formatted_harga_postdiskon.replace(",",".")}\n\n')

    total_pendapatan_toko += after_diskon


print(f'\n\nTotal Pendapatan Toko: {format_angka(total_pendapatan_toko).replace(",",".")}')
```

**Task 2 — Cari pelanggan dengan tagihan tertinggi**
1. Setelah Task 1, simpan total tagihan setiap pelanggan dalam dictionary baru bernama tagihan_pelanggan.
2. Gunakan loop untuk mencari siapa pelanggan dengan tagihan tertinggi tanpa menggunakan fungsi max() atau sort().
3. Cetak nama pelanggan tersebut beserta nilai tagihannya.

```python
tagihan_pelanggan = {}

def isi_tagihan(transaksi_perorang, total_tagihan):
    i = 0
    for data in transaksi_perorang:
        tagihan_pelanggan[data.get("nama")] = total_tagihan[i]
        i+=1

def tampilkan_tagihan(hasil):
    print('Data Tagihan:')
    for k,v in hasil.items():
        print(f'{k}: {format_angka(v).replace(",",".")}')

def sortir_tagihan():
    sorted_dictionary = sorted(tagihan_pelanggan.items(), key=lambda x: x[1],reverse=True)
    tagihan_sorted = dict(sorted_dictionary)
    return tagihan_sorted

def max_tagihan():
    max_key= max(tagihan_pelanggan, key=tagihan_pelanggan.get)
    print(f'\n{max_key} memiliki tagihan terbanyak dengan total {format_angka(tagihan_pelanggan[max_key]).replace(",",".")}')

isi_tagihan(transaksi, tagihan)
tampilkan_tagihan(sortir_tagihan())
max_tagihan()
```

**Task 3 — Simulasi antrian dengan while loop**
<br>Gunakan setup berikut:

```python
antrian = ["Budi", "Sari", "Dian", "Reza", "Ayu"]
stok_kopi = 3
```

Item kopi dalam menu adalah: Espresso, Latte, Cappuccino.
1. Gunakan while loop yang berjalan selama antrian masih ada DAN stok kopi lebih dari 0.
2. Ambil pelanggan pertama dari antrian di setiap iterasi.
3. Cek apakah pelanggan tersebut memesan item kopi. Kalau iya, kurangi stok_kopi sebesar jumlah item kopi yang dipesan.
4. Cetak status per iterasi: Melayani: Budi | Stok kopi tersisa: 2
5. Jika loop berhenti karena stok habis, cetak: Stok kopi habis. Pelanggan berikutnya tidak dapat dilayani.
6. Jika loop berhenti karena antrian kosong, cetak: Semua pelanggan telah dilayani.

```python
i = 0
kopi = ["Espresso", "Latte", "Cappuccino"]
while i < len(antrian) and stok_kopi > 0:
    nama_pelanggan = antrian[i]
    for person in transaksi:
        if person.get("nama") == nama_pelanggan:
            jumlah_pesanan = 0
            for produk in person["pesanan"]:
                if produk in kopi:
                    jumlah_pesanan += 1
                else:
                    continue

            if stok_kopi > 0:
                if stok_kopi > jumlah_pesanan:
                    stok_kopi -= jumlah_pesanan
                else:
                    stok_kopi = 0
            
            print(f'Melayani: {nama_pelanggan} | Stok kopi tersisa: {stok_kopi}')
    i+=1
    if stok_kopi == 0:
        print("Stok kopi habis. Pelanggan berikutnya tidak dapat dilayani")
        break
    elif i == len(antrian):
        print("Semua pelanggan telah dilayani")

if stok_kopi > 0:
    print(f"Stok kopi masih tersisa {stok_kopi}")
```
