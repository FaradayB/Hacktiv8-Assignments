Gunakan data berikut sebagai titik awal di notebook Anda:


**Task 1 — Informasi produk terformat:**
1. Iterasi seluruh list produk menggunakan for loop (materi ini disinggung di sesi 4 sebagai preview, Anda boleh menggunakannya).
2. Untuk setiap produk, cetak baris output dengan format:<br> 
[KATEGORI] Nama Produk | Harga: Rp X | Stok: Y unit <br>
Contoh: [LAPTOP] Laptop ASUS A516 | 3 Harga: Rp 7.500.000 | Stok: 12 unit
3. Pastikan kategori dicetak dalam huruf kapital semua (gunakan method string yang sesuai).
4. Pastikan angka harga menggunakan pemisah ribuan (gunakan f-string dengan format {:,} lalu ganti koma dengan titik).

```python
import locale
def format_currency(amount):
    return 'Rp {:,}'.format(amount)

def format_currency2des(amount):
    return 'Rp {:,.2f}'.format(amount)

print("Items: ")
for a in produk:
    print(f'[{a.get("kategori").upper()}] {a.get("nama")} | Harga: {format_currency(a.get("harga")).replace(",",".")} | Stok: {a.get("stok")}')
```

**Task 2 — Kalkulasi nilai inventory**
1. Buat variabel total_nilai_inventory bertipe integer, nilai awal 0.
2. Hitung total nilai inventory dengan cara mengalikan harga dan stok tiap produk, lalu jumlahkan semuanya.
3. Cetak hasilnya dalam format: Total Nilai Inventory: Rp X
4. Buat juga variabel rata_harga yang menyimpan rata-rata harga produk (hasil pembagian harus bertipe float). Cetak dengan 2 angka desimal.

```python
total_nilai_inventory = 0
stok = 0
for a in produk:
    total_nilai_inventory = total_nilai_inventory + (a.get("harga")*a.get("stok"))
    stok = stok + a.get("stok")


print(f'Total Nilai Inventory: {format_currency(total_nilai_inventory).replace(",",".")}')
print(f'Rata-rata Harga: {format_currency2des(total_nilai_inventory/stok).replace(",",".")}')
```

**Task 4 — Ringkasan per kategori**
1. Buat dictionary kosong bernama ringkasan_kategori.
2. Isi dictionary tersebut sehingga setiap key adalah nama kategori, dan value-nya adalah jumlah produk dalam kategori tersebut.
3. Cetak isi dictionary tersebut baris per baris dalam format: Kategori [X]: Y produk
4. Tambahkan satu produk baru ke list produk: nama: "Webcam Logitech C920", harga: 750000, stok: 15, kategori: "aksesoris". Update ringkasan_kategori setelah penambahan ini, lalu cetak hasilnya kembali.

```python
def create_dictionary():
    ringkasan_kategori = {}

    for a in produk:
        kategori = a["kategori"]
        ringkasan_kategori[kategori] = ringkasan_kategori.get(kategori, 0) + 1
    
    return ringkasan_kategori
    
def tampil_ringkasan(ringkasan_kategori: dict):
    for k, v in ringkasan_kategori.items():
        print(f'{k}: {v}')

def tambah_produk(produk_list, nama, harga, stok, kategori):
    for p in produk_list:
        if p["nama"] == nama:
            print(f"{nama}")
            return

    new_product = {
        "nama": nama,
        "harga": harga,
        "stok": stok,
        "kategori": kategori 
    }

    produk_list.append(new_product)

ringkasan = create_dictionary()
print("Ringkasan Awal:")
tampil_ringkasan(ringkasan)

tambah_produk(
    produk,
    "Webcam Logitech C920",
    750000,
    15,
    "aksesoris"
)

ringkasan = create_dictionary()
print("\nSetelah ditambah:")
tampil_ringkasan(ringkasan)
```
