**Task 3 — Groupby dan agregasi**

1. Hitung survival rate (persen penumpang yang selamat) per kelas kabin (pclass). Tampilkan hasilnya sebagai DataFrame dengan kolom pclass, jumlah_penumpang, jumlah_selamat, dan survival_rate (dalam persen, 2 desimal).
2. Gunakan groupby dengan agg() untuk menghitung rata-rata usia dan rata-rata tarif (fare) sekaligus, dikelompokkan berdasarkan pclass dan sex. Tampilkan hasilnya.
3. Dari kolom kelompok_usia yang dibuat di Task 2, hitung jumlah penumpang yang selamat dan tidak selamat per kelompok usia. Tampilkan sebagai tabel pivot menggunakan pd.crosstab() atau groupby.

````python
#Task 3.1
survive_rate = df.groupby('pclass').agg(
    jumlah_selamat=('survived', 'sum'),
    jumlah_penumpang=('survived', 'count') 
).reset_index()

survive_rate['survival_rate'] = (survive_rate['jumlah_selamat'] / survive_rate['jumlah_penumpang']*100).round(2)

survive_rate
````

````python
#Task 3.2
fare_age = df.groupby(['pclass', 'sex']).agg(
    avg_age=('age', 'mean'),
    avg_fare=('fare', 'mean') 
).round(2).reset_index()

fare_age
````
